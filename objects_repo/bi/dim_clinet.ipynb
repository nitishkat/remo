```python
%sql
CREATE or replace TABLE   corpadmin.bigold.dim_client (
  client_sk          BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key: new value generated for every row (one per client version)',
  clientid           STRING NOT NULL COMMENT 'Natural / business key from the source system',
  clientname         STRING COMMENT 'Client name',
  clienttype         STRING COMMENT 'Client type (e.g. individual, corporate)',
  address            STRING COMMENT 'Street address',
  city               STRING COMMENT 'City',
  country            STRING COMMENT 'Country',
  start_ts TIMESTAMP NOT NULL COMMENT 'SCD2: version valid from this timestamp',
  end_ts   TIMESTAMP COMMENT 'SCD2: version valid until this timestamp; NULL for the current row',
  is_current         BOOLEAN NOT NULL COMMENT 'SCD2: TRUE for the current version of the client',
  CONSTRAINT client_pk PRIMARY KEY (client_sk),
  CONSTRAINT one_current_per_client UNIQUE (clientid, is_current)
)
CLUSTER BY (clientid)
COMMENT 'Type 2 SCD dimension: full history of client versions. Current row: is_current = true and effective_end_ts = NULL.'
```
