```python
%sql
CREATE OR REPLACE TABLE corpadmin.odssilver.client (
  client_sk          BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key: new value generated for every row (one per client version)',
  clientid           STRING NOT NULL COMMENT 'Natural / business key from the source system',
  clientname         STRING COMMENT 'Client name',
  clienttype         STRING COMMENT 'Client type (e.g. individual, corporate)',
  effective_start_ts TIMESTAMP NOT NULL COMMENT 'SCD2: version valid from this timestamp',
  effective_end_ts   TIMESTAMP COMMENT 'SCD2: version valid until this timestamp; NULL for the current row',
  CONSTRAINT client_pk PRIMARY KEY (client_sk)
)
CLUSTER BY (clientid)
COMMENT 'SCD Type 2 client table (no is_current column): full history of client versions. Current row: effective_end_ts IS NULL.'
```

```python
%sql
CREATE OR REPLACE  table corpadmin.odssilver.clientaddress (
  clientaddress_sk   BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Surrogate key: new value generated for every row (one per address version)',
  clientid           STRING NOT NULL COMMENT 'Natural / business key, joins to corpadmin.bigold.client.clientid',
  address            STRING COMMENT 'Street address',
  city               STRING COMMENT 'City',
  country            STRING COMMENT 'Country',
  start_ts          TIMESTAMP NOT NULL COMMENT 'SCD2: version valid from this timestamp',
  end_ts            TIMESTAMP COMMENT 'SCD2: version valid until this timestamp; NULL for the current row',
  CONSTRAINT ods_clientaddress_pk PRIMARY KEY (clientaddress_sk)
)
CLUSTER BY (clientid)
COMMENT 'SCD Type 2 client address table (no is_current column): full history of address versions. Current row: end_ts IS NULL.'
```
